```python
# =========================
# PARAGUAY — WF1
# =========================

PARAGUAY_ASUNCION = ["Asunción"]
PARAGUAY_SAN_PEDRO = ["San Pedro"]
PARAGUAY_CENTRAL = ["Central"]
PARAGUAY_CAAGUAZU = ["Caaguazú"]
PARAGUAY_ALTO_PARAGUAY = ["Alto Paraguay"]
PARAGUAY_BOQUERON = ["Boquerón"]
PARAGUAY_ALTO_PARANA = ["Alto Paraná"]
PARAGUAY_ITAPUA = ["Itapúa"]
PARAGUAY_RESTO = ["Amambay","Caazapá","Canindeyú","Concepción","Cordillera","Guairá","Misiones","Ñeembucú","Paraguarí","Presidente Hayes"]

# Validation:
# All 18 GADM level-1 units accounted for exactly once.

# =========================
# Bangladesh (WF1)
# =========================

BANGLADESH_BARISHAL = ["Barisal"]
BANGLADESH_CHATTOGRAM = ["Chittagong"]
BANGLADESH_DHAKA = ["Dhaka"]  # will later subtract Mymensingh districts
BANGLADESH_KHULNA = ["Khulna"]
BANGLADESH_MYMENSINGH_DISTRICTS = ["Jamalpur","Nasirabad","Netrakona","Sherpur"]
BANGLADESH_RAJSHAHI = ["Rajshahi"]
BANGLADESH_RANGPUR = ["Rangpur"]
BANGLADESH_SYLHET = ["Sylhet"]

# Validation:
# - All Level 1 names match GADM exactly
# - Mymensingh does not exist in GADM Level 1
# - Mymensingh must therefore be constructed from Level 2 districts inside Dhaka
# - Dhaka final region must exclude these four districts



# =========================
# Togo (WF1)
# =========================

TOGO_SAVANES = ["Savanes"]
TOGO_KARA = ["Kara"]
TOGO_CENTRALE = ["Centre"]
TOGO_PLATEAUX = ["Plateaux"]
TOGO_MARITIME = ["Maritime"]

TOGO_LOME_GOLFE = ["Golfe (incl Lomé)"]

# Validation:
# - All Level 1 names match GADM
# - "Centre" in GADM corresponds to survey "Centrale"
# - Lomé Commune + Golfe Urbain are merged into one region
# - "Golfe (incl Lomé)" is a Level 2 unit inside Maritime
# - Maritime final region must exclude this Level 2 geometry


# Part 2 =========================--------------------------------------------------------------------------------------------


# 
# 
#  =========================
# PARAGUAY — WF2
# =========================

PARAGUAY_ASUNCION = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_ASUNCION))
PARAGUAY_SAN_PEDRO = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_SAN_PEDRO))
PARAGUAY_CENTRAL = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_CENTRAL))
PARAGUAY_CAAGUAZU = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_CAAGUAZU))
PARAGUAY_ALTO_PARAGUAY = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_ALTO_PARAGUAY))
PARAGUAY_BOQUERON = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_BOQUERON))
PARAGUAY_ALTO_PARANA = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_ALTO_PARANA))
PARAGUAY_ITAPUA = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_ITAPUA))
PARAGUAY_RESTO = GADM_level1_prep.filterMetadata('NAME_0','equals','Paraguay').filter(ee.Filter.inList("NAME_1",PARAGUAY_RESTO))




# =========================
# Bangladesh (WF2)
# =========================

BANGLADESH_BARISHAL = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_BARISHAL))
BANGLADESH_CHATTOGRAM = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_CHATTOGRAM))
BANGLADESH_DHAKA_FULL = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_DHAKA))
BANGLADESH_KHULNA = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_KHULNA))
BANGLADESH_RAJSHAHI = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_RAJSHAHI))
BANGLADESH_RANGPUR = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_RANGPUR))
BANGLADESH_SYLHET = GADM_level1_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(ee.Filter.inList("NAME_1",BANGLADESH_SYLHET))

# Mymensingh districts from Level 2
BANGLADESH_MYMENSINGH = GADM_level2_prep.filterMetadata('NAME_0','equals','Bangladesh').filter(
    ee.Filter.inList("NAME_2", BANGLADESH_MYMENSINGH_DISTRICTS)
)

# Dhaka minus Mymensingh districts
BANGLADESH_DHAKA = ee.FeatureCollection([
    ee.Feature(
        BANGLADESH_DHAKA_FULL.geometry().difference(
            BANGLADESH_MYMENSINGH.geometry(),
            10
        )
    )
]).map(lambda f: f.set('NAME_0','Bangladesh')
                  .set('NAME_1','DHAKA')
                  .set('HH7','BANGLADESH_DHAKA'))



# Togo (WF2)
# =========================

TOGO_SAVANES = GADM_level1_prep.filterMetadata('NAME_0','equals','Togo').filter(ee.Filter.inList("NAME_1",TOGO_SAVANES))
TOGO_KARA = GADM_level1_prep.filterMetadata('NAME_0','equals','Togo').filter(ee.Filter.inList("NAME_1",TOGO_KARA))
TOGO_CENTRALE = GADM_level1_prep.filterMetadata('NAME_0','equals','Togo').filter(ee.Filter.inList("NAME_1",TOGO_CENTRALE))
TOGO_PLATEAUX = GADM_level1_prep.filterMetadata('NAME_0','equals','Togo').filter(ee.Filter.inList("NAME_1",TOGO_PLATEAUX))
TOGO_MARITIME_FULL = GADM_level1_prep.filterMetadata('NAME_0','equals','Togo').filter(ee.Filter.inList("NAME_1",TOGO_MARITIME))

TOGO_LOME_GOLFE = GADM_level2_prep.filterMetadata('NAME_0','equals','Togo').filter(
    ee.Filter.inList("NAME_2", TOGO_LOME_GOLFE)
)

# Maritime minus Lomé/Golfe
TOGO_MARITIME = ee.FeatureCollection([
    ee.Feature(
        TOGO_MARITIME_FULL.geometry().difference(
            TOGO_LOME_GOLFE.geometry(),
            10
        )
    )
]).map(lambda f: f.set('NAME_0','Togo')
                  .set('NAME_1','MARITIME')
                  .set('HH7','TOGO_MARITIME'))

#
# 

# Part 3======================================================================================0


Paraguay = ee.FeatureCollection([
    ee.Feature(PARAGUAY_ASUNCION.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','ASUNCION').set('NAME_1s',PARAGUAY_ASUNCION.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_ASUNCION.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_ASUNCION'),
    ee.Feature(PARAGUAY_SAN_PEDRO.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','SAN PEDRO').set('NAME_1s',PARAGUAY_SAN_PEDRO.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_SAN_PEDRO.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_SAN_PEDRO'),
    ee.Feature(PARAGUAY_CENTRAL.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','CENTRAL').set('NAME_1s',PARAGUAY_CENTRAL.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_CENTRAL.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_CENTRAL'),
    ee.Feature(PARAGUAY_CAAGUAZU.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','CAAGUAZU').set('NAME_1s',PARAGUAY_CAAGUAZU.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_CAAGUAZU.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_CAAGUAZU'),
    ee.Feature(PARAGUAY_ALTO_PARAGUAY.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','ALTO PARAGUAY').set('NAME_1s',PARAGUAY_ALTO_PARAGUAY.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_ALTO_PARAGUAY.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_ALTO_PARAGUAY'),
    ee.Feature(PARAGUAY_BOQUERON.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','BOQUERON').set('NAME_1s',PARAGUAY_BOQUERON.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_BOQUERON.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_BOQUERON'),
    ee.Feature(PARAGUAY_ALTO_PARANA.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','ALTO PARANA').set('NAME_1s',PARAGUAY_ALTO_PARANA.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_ALTO_PARANA.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_ALTO_PARANA'),
    ee.Feature(PARAGUAY_ITAPUA.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','ITAPUA').set('NAME_1s',PARAGUAY_ITAPUA.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_ITAPUA.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_ITAPUA'),
    ee.Feature(PARAGUAY_RESTO.union().geometry()).set('NAME_0','Paraguay').set('NAME_1','RESTO').set('NAME_1s',PARAGUAY_RESTO.aggregate_array('NAME_1').join('|')).set('GID_1s',PARAGUAY_RESTO.aggregate_array('GID_1').join('|')).set('HH7','PARAGUAY_RESTO')
])



Bangladesh = ee.FeatureCollection([
    ee.Feature(BANGLADESH_BARISHAL.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','BARISHAL').set('NAME_1s',BANGLADESH_BARISHAL.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_BARISHAL.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_BARISHAL'),
    ee.Feature(BANGLADESH_CHATTOGRAM.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','CHATTOGRAM').set('NAME_1s',BANGLADESH_CHATTOGRAM.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_CHATTOGRAM.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_CHATTOGRAM'),
    ee.Feature(BANGLADESH_DHAKA.geometry()).set('NAME_0','Bangladesh').set('NAME_1','DHAKA').set('HH7','BANGLADESH_DHAKA'),
    ee.Feature(BANGLADESH_MYMENSINGH.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','MYMENSINGH').set('NAME_2s',BANGLADESH_MYMENSINGH.aggregate_array('NAME_2').join('|')).set('GID_2s',BANGLADESH_MYMENSINGH.aggregate_array('GID_2').join('|')).set('HH7','BANGLADESH_MYMENSINGH'),
    ee.Feature(BANGLADESH_KHULNA.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','KHULNA').set('NAME_1s',BANGLADESH_KHULNA.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_KHULNA.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_KHULNA'),
    ee.Feature(BANGLADESH_RAJSHAHI.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','RAJSHAHI').set('NAME_1s',BANGLADESH_RAJSHAHI.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_RAJSHAHI.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_RAJSHAHI'),
    ee.Feature(BANGLADESH_RANGPUR.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','RANGPUR').set('NAME_1s',BANGLADESH_RANGPUR.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_RANGPUR.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_RANGPUR'),
    ee.Feature(BANGLADESH_SYLHET.union().geometry()).set('NAME_0','Bangladesh').set('NAME_1','SYLHET').set('NAME_1s',BANGLADESH_SYLHET.aggregate_array('NAME_1').join('|')).set('GID_1s',BANGLADESH_SYLHET.aggregate_array('GID_1').join('|')).set('HH7','BANGLADESH_SYLHET')
])



Togo = ee.FeatureCollection([
    ee.Feature(TOGO_SAVANES.union().geometry()).set('NAME_0','Togo').set('NAME_1','SAVANES').set('NAME_1s',TOGO_SAVANES.aggregate_array('NAME_1').join('|')).set('GID_1s',TOGO_SAVANES.aggregate_array('GID_1').join('|')).set('HH7','TOGO_SAVANES'),
    ee.Feature(TOGO_KARA.union().geometry()).set('NAME_0','Togo').set('NAME_1','KARA').set('NAME_1s',TOGO_KARA.aggregate_array('NAME_1').join('|')).set('GID_1s',TOGO_KARA.aggregate_array('GID_1').join('|')).set('HH7','TOGO_KARA'),
    ee.Feature(TOGO_CENTRALE.union().geometry()).set('NAME_0','Togo').set('NAME_1','CENTRALE').set('NAME_1s',TOGO_CENTRALE.aggregate_array('NAME_1').join('|')).set('GID_1s',TOGO_CENTRALE.aggregate_array('GID_1').join('|')).set('HH7','TOGO_CENTRALE'),
    ee.Feature(TOGO_PLATEAUX.union().geometry()).set('NAME_0','Togo').set('NAME_1','PLATEAUX').set('NAME_1s',TOGO_PLATEAUX.aggregate_array('NAME_1').join('|')).set('GID_1s',TOGO_PLATEAUX.aggregate_array('GID_1').join('|')).set('HH7','TOGO_PLATEAUX'),
    ee.Feature(TOGO_MARITIME.geometry()).set('NAME_0','Togo').set('NAME_1','MARITIME').set('HH7','TOGO_MARITIME'),
    ee.Feature(TOGO_LOME_GOLFE.union().geometry()).set('NAME_0','Togo').set('NAME_1','LOME_GOLFE_URBAIN').set('NAME_2s',TOGO_LOME_GOLFE.aggregate_array('NAME_2').join('|')).set('GID_2s',TOGO_LOME_GOLFE.aggregate_array('GID_2').join('|')).set('HH7','TOGO_LOME_GOLFE')

])



GADM_level1 = GADM_level1.merge(Paraguay).merge(Bangladesh).merge(Togo)
```
